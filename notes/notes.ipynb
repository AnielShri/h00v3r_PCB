# Oscilator selection

## HSE

Chosen component => [**X322525MOB4SI**](https://jlcpcb.com/partdetail/Yxc_CrystalOscillators-X322525MOB4SI/C9006) \
Description => 	25MHz SMD crystal oscillator 12pF ±10ppm ±20ppm SMD3225-4P Crystals ROHS \
Relevant info => 
* $ C_L = 12pF $
* $ ESR = 50\Omega $
* $ C_0 = 3pF \ Max $

## Calculating $ C_{L1,2} $
From [AN2867](https://www.st.com/resource/en/application_note/cd00221665-oscillator-design-guide-for-stm8af-al-s-stm32-mcus-and-mpus-stmicroelectronics.pdf) (section 3.3, 3.5) \
Selecting the oscilator capacitors: 
$$ C_L = \frac{C_{L1} * C_{L2}}{C_{L1} + C_{L2}} + C_s $$ 
Choosing $ C_{L1} = C_{L2} = C_{L12} $ and assuming $ C_s > C_L $ => 
$$ C_{L12} = 2 * (C_L - C_s) $$
Selecting external resistor: 
$$ R_{ext} = \frac{1}{2 * \pi * F * C_{L2}} $$ 

Validate with either:
* $ G_{m\_crit\_max} > g_{mcrit} $ if $ G_{m\_crit\_max} $ is specifed
* $ gain_{margin} > 5 $ if $ g_m $ is specified

Where $$ g_{mcrit} = 4 * (ESR + R_{ext}) * (2 \pi F)^2 * (C_0 + C_L)^2 $$
And $$ gain_{margin} = \frac{g_m}{g_{mcrit}} $$

From [AN2586](https://www.st.com/content/ccc/resource/technical/document/application_note/6c/a3/24/49/a5/d4/4a/db/CD00164185.pdf/files/CD00164185.pdf/jcr:content/translations/en.CD00164185.pdf) Rev 8 (section 3.1) 
* Typical value of $ R_{ext} $ is in the range of 5 to 6 $ ESR $ 
* $ C_s $ is typically between 2 and 7 pF

```python
import math

f_HSE = 25e6 
C_L = 12e-12
C_s = 4.5e-12
C_0 = 3e-12
ESR = 50

C_L12 = 2 * (C_L - C_s)
R_ext = 1/(2 * math.pi * f_HSE * C_L12)
R_ext_alt = 6 * ESR
R_ext_chosen = R_ext_alt

Gm_crit_max = 1.5
gm_crit = 4 * (ESR + R_ext_chosen) * math.pow(2*math.pi*f_HSE, 2) * math.pow(C_0 + C_L, 2)


print("C_L12: {:.2f}pF".format(C_L12 * 1e12))
print("R_ext: {:.2f} or {:.2f}".format(R_ext, R_ext_alt))
print("GM_crit_max > gm_crit => {:.2f} >> {:.2f}".format(Gm_crit_max, gm_crit))

```
